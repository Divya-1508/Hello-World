```python
# 1. Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime

# Create sample CSV dataset inside the code
data = {
    "ID": [1, 2, 3, 4, 5, 5, 6, 7],
    "Name": ["Rahul", "ANJALI", "rohan", "Priya", "Aman", "Aman", "NEHA", "Raj"],
    "City": ["Pune", "mumbai", "PUNE", None, "Delhi", "Delhi", "mumbai", "Pune"],
    "Date_of_Birth": [
        "2003-05-10",
        "2002-08-15",
        "2004-01-20",
        "2003-11-05",
        "2001-03-12",
        "2001-03-12",
        "2002-07-25",
        "1995-06-10"
    ],
    "Salary": [30000, 35000, 32000, None, 40000, 40000, 36000, 200000]
}

df = pd.DataFrame(data)
df.to_csv("employee_data.csv", index=False)


# 2. Load the CSV dataset
df = pd.read_csv("employee_data.csv")

print("Original Dataset:")
print(df)


# 3. Display dataset information
print("\nDataset Information:")
print(df.info())

print("\nDataset Description:")
print(df.describe())


# 4. Identify missing values
print("\nMissing Values:")
print(df.isnull().sum())


# 5. Fill missing values
df["City"] = df["City"].fillna("Unknown")
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print("\nDataset after filling missing values:")
print(df)


# 6. Remove duplicate records
df = df.drop_duplicates()

print("\nDataset after removing duplicates:")
print(df)


# 7. Standardize text values
df["Name"] = df["Name"].str.title()
df["City"] = df["City"].str.title()

print("\nDataset after standardizing text:")
print(df)


# 8. Convert date of birth into age
df["Date_of_Birth"] = pd.to_datetime(df["Date_of_Birth"])

current_year = datetime.now().year
df["Age"] = current_year - df["Date_of_Birth"].dt.year

print("\nDataset after calculating Age:")
print(df)


# 9. Detect outliers using a box plot
plt.figure(figsize=(7, 5))
plt.boxplot(df["Salary"])
plt.title("Salary Outlier Detection")
plt.ylabel("Salary")
plt.show()


# 10. Generate visual reports

# Salary distribution
plt.figure(figsize=(7, 5))
plt.hist(df["Salary"], bins=5)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.show()

# City-wise employee count
plt.figure(figsize=(7, 5))
df["City"].value_counts().plot(kind="bar")
plt.title("Employees by City")
plt.xlabel("City")
plt.ylabel("Number of Employees")
plt.show()


# 11. Save the cleaned dataset
df.to_csv("cleaned_employee_data.csv", index=False)

print("\nCleaned dataset saved successfully!")
print("File name: cleaned_employee_data.csv")
```
